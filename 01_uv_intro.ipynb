# Intro

### Install uv

- On macOS and Linux:  
curl -LsSf https://astral.sh/uv/install.sh | sh   
  
- On Windows (from terminal):  
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"

## Basic usage

### Generate an environment  
- uv venv --python 3.12

### Activate an environment
macOS and Linux:  
- source .venv/bin/activate
  
Windows:  
- .venv\Scripts\activate

### Install dependencies

- uv pip install requests

---

## Advanced usage

### Inbitialize a project

- uv init myproject  
- cd myproject

myproject/  
├── .git/              # yes, it inits a git repo too
├── .gitignore  
├── .python-version  
├── pyproject.toml  
├── README.md  
└── main.py  

### Pin python verion

- uv python pin 3.12

### Add dependency

- uv add requests

### Remove dependency

- uv remove requests

### Run things

- uv run main.py  
- uv run python

No need to activate env. This gets activated automatically.
