```python
import random
choices = ["rock", "paper", "scissors"]
computer_choice = random.choice(choices)

your_choice = "paper"

if your_choice == computer_choice:
    print(f"Both chose {your_choice}. It is a tie!")
elif your_choice == "scissors" and computer_choice == "paper":
    print(f"You chose {your_choice} and computer chose {computer_choice}. You win!")
elif your_choice == "paper" and computer_choice == "rock":
    print(f"You chose {your_choice} and computer chose {computer_choice}. You win!")
elif your_choice == "rock" and computer_choice == "scissors":
    print(f"You chose {your_choice} and computer chose {computer_choice}. You win!")
else:
    print(f"You chose {your_choice} and computer chose {computer_choice}. You lose!")
```

```python
import random

def roll_until_six(sides):
    dice = 1
    rolls = 0
    
    while dice != 6:
        dice = random.randint(1,sides)
        rolls = rolls + 1
        
    return rolls
    
```

```python
dices = {"cube": 6, "octahedron": 8, "dodecahedron": 10, "icosahedron": 20}
```

```python
for name, sides in dices.items():
    print(f"{name.capitalize()} needed {roll_until_six(sides)} to get a 6.")
```
