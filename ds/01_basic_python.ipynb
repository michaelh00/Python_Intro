## Part 0 — How a piece of Python code is actually put together

A `#` starts a comment. Everything after it, on that line, is ignored by Python entirely — it's there for whoever reads the code next, which is often the same person a few months later.

```
# This whole line is a comment and does nothing when run
print("Documents downloaded today: 3")  # a comment can also follow real code on the same line
```

Indentation is not a style preference in Python, it's part of the syntax. Anything that should run "inside" an `if`, a loop, or a function needs to be indented — consistently, by convention four spaces — and Python uses that indentation, not brackets or an `end` keyword, to know where a block starts and stops.

```
tariff_rate = 12.5

if tariff_rate > 10:
    print("above the general threshold")
    print("this line is inside the if-block too")

print("this line is not indented, so it always runs")
```

Try it: delete the four spaces in front of one of the two indented `print` lines above and re-run the cell. That produces an `IndentationError` — worth seeing once deliberately, since it's one of the most common early mistakes and the error message is unusually literal about what's wrong.

A single statement can be spread across multiple lines by wrapping it in parentheses, and, separately, multiple short statements can share a single line using a semicolon — both used far less often than plain one-statement-per-line code, but worth recognising when they show up.

```
total = (12.5
         + 7.0
         + 3.25)
print(total)

a = 5; b = 3; print(a - b)
```

Parentheses do two different jobs, and it's worth being able to tell them apart on sight. After a name, they *call* whatever that name refers to — running a function. On their own, around an expression, they *group* it, controlling the order operations happen in, same as in ordinary arithmetic.

```
print(sorted([3, 1, 2]))     # calling: this runs the sorted() function
value = (3 + 4) * 2           # grouping: forces the addition to happen first
print(value)

print(sorted)                 # no parentheses -- this is just the function itself, not a call to it
```

## Part 1 — What a variable actually is, then types, operators, and errors

It's tempting to picture a variable as a labelled box holding a value. That model works for a while, but it isn't what's actually happening, and the real version explains things the box model can't. A variable is a name bound to an object living in memory — closer to a label stuck onto something than a container holding it.

```
filename = "WT-DS316-AB-R.txt"
case_number = 316
is_appellate_body = True
tariff_rate = 5.0

print(type(filename))
print(type(case_number))
print(type(is_appellate_body))
print(type(tariff_rate))
```

Almost everything in Python is an object, meaning it carries both a value and a set of things it knows how to do, accessed with a dot. Two kinds of dotted things exist: attributes, which are just stored information, and methods, which are functions belonging to the object — the giveaway is that a method needs parentheses to actually run, even if there's nothing inside them.

```
tariff_rate = 5.5
print(tariff_rate.is_integer())   # method (note the parentheses): does this float have no fractional part?
print(tariff_rate.real)           # attribute (no parentheses): the value itself

filename = "WT-DS316-AB-R.txt"
print(filename.upper())           # method: returns a new string
print(filename)                   # the original is unchanged -- filename.upper() didn't modify it in place
```

That last point matters more than it might seem: calling a method doesn't change the object unless the method specifically says it does, and most string and number methods don't. This comes up constantly with strings, which is where Part 2 picks up.

Converting between types deliberately, rather than hoping Python guesses correctly, uses `str()`, `int()`, and `float()`.

```
case_number_as_text = str(case_number)
tariff_rate_as_int = int(tariff_rate)

print(case_number_as_text, type(case_number_as_text))
print(tariff_rate_as_int, type(tariff_rate_as_int))
```

Arithmetic covers more than the basic four operators — `//` and `%` come up constantly once you're doing anything with counts or batches, and comparisons combine with `and`, `or`, `not` to build up more specific conditions than a single comparison can express.

```
a = 17
b = 5

print(a + b, a - b, a * b)
print(a / b)    # true division: always gives a float
print(a // b)   # floor division: whole number of times b fits into a
print(a % b)    # remainder after floor division
print(a ** 2)   # power

print(a > b, a == b, a != b)
print(a > 10 and b < 10)
print(a > 100 or b < 10)
print(not is_appellate_body)
```

Now something that's supposed to fail. Concatenating a string and an integer with `+` doesn't work — Python won't silently convert one to match the other, which is a deliberate design choice, not a bug to work around. Run it and read what comes back.

```
case_number = 316
message = "Case number: " + case_number  # this line is broken on purpose
print(message)
```

The bottom line of that traceback is the one that matters first: `TypeError: can only concatenate str (not "int") to str`. The lines above it show where in the code things went wrong, read top to bottom; the actual error, read bottom to top, is what to look at first. This shape — error type, then a description, with the call stack above it — is the same for every error type in Python, so it's worth getting comfortable reading it now rather than treating every red box as a dead end.

The fix here is exactly what the earlier type-conversion cell was practicing:

```
message = "Case number: " + str(case_number)
print(message)
```

## Part 2 — Strings, in depth

A filename is a string, and everything here applies directly to the documents in `sample_data/`. As just seen with `.upper()`, string methods don't change the original — they return a new one, which is why they're usually chained or reassigned rather than called and ignored.

```
name = "  Costa Rica - Tariff Schedule (revised) FINAL v3.csv  "

print(repr(name))                          # repr() shows the leading/trailing spaces explicitly
print(repr(name.strip()))
print(name.strip().lower())
print(name.strip().replace(" ", "_"))
print(name.strip().endswith(".csv"))
print(name.strip().startswith("Costa"))
```

`.split()` breaks a string apart wherever a character appears, returning a list; `.join()` does the reverse, gluing a list of strings back together with a separator of your choice. Together they're how you get from a filename to its parts and back.

```
symbol = "WT-DS316-AB-R"
parts = symbol.split("-")
print(parts)
print(len(parts))

rejoined = "/".join(parts)
print(rejoined)  # this is the actual WTO document symbol format: WT/DS316/AB/R
```

f-strings also format numbers, not just insert them as-is — useful the moment a tariff rate needs a fixed number of decimal places instead of whatever Python happens to produce.

```
tariff_rate = 5.0
print(f"{tariff_rate:.2f}")     # two decimal places
print(f"{tariff_rate:6.2f}")    # padded to a minimum width of 6

doc_type = "panel_report"
case_number = 412
year = 2019
built_name = f"{doc_type}_{case_number}_{year}.txt"
print(built_name)
```

## Part 3 — Lists, dictionaries, tuples, sets, and unpacking

Lists have methods that modify them in place: `.append()` adds an item, `.sort()` reorders the list itself. `sorted()`, by contrast, is a function that returns a new sorted list and leaves the original untouched — a distinction worth keeping straight, since mixing them up is a common source of confusion later.

```
case_numbers = [316, 135, 412]
case_numbers.append(58)
print(case_numbers)

case_numbers.sort()
print(case_numbers)                          # sorted in place, ascending

print(sorted(case_numbers, reverse=True))     # a new list, descending — case_numbers itself is unchanged
print(case_numbers)

print(len(case_numbers))
print(316 in case_numbers)
print(999 in case_numbers)
```

Dictionaries have equivalents for looking at their contents as a whole: `.keys()`, `.values()`, and `.items()` (pairs of both together). We'll loop over `.items()` properly next session — for now, just see what it looks like.

```
document = {"filename": "WT-DS316-AB-R.txt", "case_number": 316, "doc_type": "appellate_body"}

print(document.keys())
print(document.values())
print(document.items())
```

Two more structures worth knowing, even briefly: a tuple is an ordered, fixed group of values, good for something like a pair of coordinates that shouldn't accidentally grow a third item. A set holds unique values only, the natural tool for questions about how many distinct case numbers are actually present.

```
bern_coordinates = (46.94, 7.44)
print(bern_coordinates[0], bern_coordinates[1])

case_numbers_with_duplicates = [316, 135, 316, 412, 135]
unique_cases = set(case_numbers_with_duplicates)
print(unique_cases)
print(len(case_numbers_with_duplicates), "entries,", len(unique_cases), "unique")
```

Assigning to more than one variable at once — unpacking — is worth naming explicitly, because it shows up constantly from here on, including next session when functions hand back more than one result at a time.

```
case_number, doc_type = 316, "appellate_body"
print(case_number, doc_type)

lat, lon = bern_coordinates
print(lat, lon)

# swapping two values, using unpacking, with no temporary variable needed
x, y = 1, 2
x, y = y, x
print(x, y)
```

Putting it together: a list of dictionaries, one per document, is the structure the rest of the course builds on.

```
documents = [
    {"filename": "WT-DS316-AB-R.txt", "case_number": 316, "doc_type": "appellate_body", "year": None},
    {"filename": "WT-DS135-R_asbestos.txt", "case_number": 135, "doc_type": "panel_report", "year": None},
    {"filename": "G_MA_TAR_RS_70_Argentina.csv", "case_number": None, "doc_type": "tariff_schedule", "year": None},
]

case_numbers_present = []
case_numbers_present.append(documents[0]["case_number"])
case_numbers_present.append(documents[1]["case_number"])
print(case_numbers_present)
```
