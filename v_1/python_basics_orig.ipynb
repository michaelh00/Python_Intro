## How a Jupyter notebook works — thirty seconds

A notebook is a sequence of **cells**. Two kinds matter today:

- **Markdown cells** hold text — like this one.
- **Code cells** hold Python — like the next one.

Run a cell by clicking it and pressing `Shift + Enter`. Whatever the last
line of the cell produces shows up right below it. Cells share memory:
once you've defined something in one cell, later cells can use it — but
only if you've actually **run** the earlier cell. If a later cell fails
with `NameError`, that's the most common cause: you skipped ahead
without running what came before.


````python
# Click this cell, then press Shift + Enter to run it.
print("Hello from the World Trade Institute")
2 + 3
````

Two things showed up: the printed line, then the number `5`. Python
automatically displays the value of the **last expression** in a cell —
so `2 + 3` becomes visible without needing `print()`. Everything above
the last line only shows up if you explicitly `print` it.

## Part 1 — Variables, types, operators

*(Live coding — roughly 20 minutes.)*

Think of a variable as a **label stuck onto a value**. The `=` sign
attaches the label on the left to the value on the right. It is *not*
the mathematical equals sign — it does one thing, in one direction.


````python
country = "Switzerland"
gdp_trillion_usd = 0.9
year = 2024
````

Nothing printed, because assignment doesn't produce a value to display.
The three variables now exist, though. We can look at any of them:


````python
country
````

````python
print(country, "had a GDP of about", gdp_trillion_usd, "trillion USD in", year)
````

Variable names can be almost anything, but there are a few rules and a
few strong conventions:

- Start with a letter or underscore, never a digit.
- Only letters, digits, and underscores — no spaces, no hyphens.
- Case matters: `Country` and `country` are two different names.
- Use lowercase with underscores by convention: `case_number`, not
  `CaseNumber` or `casenumber`.
- Pick names that mean something to the next person reading the code
  (usually you, in six months).


````python
# Good
tariff_rate = 5.5
number_of_documents = 42

# Legal but painful
x = 5.5
tariffRate = 5.5

# Broken -- uncomment to see the SyntaxError, then re-comment
# 2nd_country = "Kenya"
# tariff rate = 5.5
````

Reassigning a variable simply moves the label to a new value. The old
value is forgotten.


````python
tariff_rate = 5.5
print(tariff_rate)

tariff_rate = 12.0
print(tariff_rate)
````

### The four everyday types

Python figures out the type from the value on the right of the `=`. You
can always ask, with `type()`.


````python
case_number = 316          # int    -- a whole number
tariff_rate = 5.5           # float  -- a number with a decimal point
country = "Kenya"           # str    -- a string, i.e. text
is_confidential = False     # bool   -- either True or False

print(type(case_number))
print(type(tariff_rate))
print(type(country))
print(type(is_confidential))
````

### Arithmetic

The obvious four plus three more that come up all the time:

| Operator | Meaning                                    |
|----------|--------------------------------------------|
| `+`      | addition                                   |
| `-`      | subtraction                                |
| `*`      | multiplication                             |
| `/`      | division (always gives a `float`)          |
| `//`     | floor division (drop the fractional part)  |
| `%`      | remainder                                  |
| `**`     | power                                      |


````python
pages = 137
chapters_of_ten_pages = pages // 10
leftover_pages = pages % 10

print("full chapters:", chapters_of_ten_pages)
print("leftover pages:", leftover_pages)
print("if we doubled the length:", pages * 2)
print("if we squared it:", pages ** 2)
````

### Comparisons return booleans

Comparisons ask a yes/no question. The answer is always `True` or
`False`, which is why booleans exist.


````python
tariff_rate = 5.5

print(tariff_rate > 10)
print(tariff_rate == 5.5)   # note the DOUBLE equals for comparison
print(tariff_rate != 0)
````

`=` assigns. `==` compares. Getting these two confused is one of the
top-three beginner mistakes — worth writing on a post-it.

You can also combine several conditions with `and`, `or`, `not`:


````python
case_number = 316
tariff_rate = 5.5

print(case_number > 300 and tariff_rate < 10)
print(case_number > 500 or tariff_rate < 10)
print(not (case_number > 500))
````

### Converting between types

Sometimes Python needs a nudge. `int()`, `float()`, and `str()` convert
a value into that type — or complain loudly if the value doesn't make
sense as that type.


````python
print(int(5.9))           # drops the decimal, does NOT round -> 5
print(float("3.14"))      # a string that happens to look like a number -> 3.14
print(str(2024))          # the number 2024 as text -> "2024"
````

### An error you should meet on purpose

Python will not silently combine a string with a number. The next cell
is broken deliberately. Run it, read the error, then look at the fix.


````python
case_number = 316
message = "Case number: " + case_number
````

Read the traceback from the **bottom up**: the last line is what went
wrong. Here it says:

```
TypeError: can only concatenate str (not "int") to str
```

Python is telling you: you tried to glue a number onto a string with
`+`, and I refuse. The fix is to convert the number first:


````python
case_number = 316
message = "Case number: " + str(case_number)
print(message)
````

## Part 2 — Strings, lists, dictionaries, indexing

*(Live coding — roughly 20 minutes.)*

Once you have more than one number or one word, you need a container.
Python has four you should meet on day one: **strings** (which we've
already used), **lists**, **dictionaries**, and — very briefly —
**tuples** and **sets**.


### Strings are more than just text

A string is a sequence of characters, and it comes with a lot of
built-in helpers. Try these one at a time.


````python
filename = "  Costa Rica - Tariff Schedule FINAL.csv  "

print(filename.strip())                 # removes leading/trailing whitespace
print(filename.lower())                 # everything lowercase
print(filename.strip().lower())         # chained: first strip, then lowercase
print(filename.replace(" ", "_"))       # replace one substring with another
print(filename.endswith(".csv"))        # True/False question
print(len(filename))                    # how many characters
````

Notice `filename` itself hasn't changed. Every string method returns a
**new** string — it never modifies the original. If you want to keep
the cleaned version, you have to assign it:


````python
filename = "  Costa Rica - Tariff Schedule FINAL.csv  "
cleaned = filename.strip().lower().replace(" ", "_")

print(filename)   # original, unchanged
print(cleaned)    # the new one we kept
````

### f-strings: putting variables into text

Concatenating strings with `+` gets ugly quickly. **f-strings** are the
clean way to build a sentence with variables inside. Prefix a string
with `f` and put variable names in `{}`.


````python
country = "Kenya"
tariff_rate = 7.5
year = 2024

# The clunky way
print("In " + str(year) + ", " + country + " had a tariff rate of " + str(tariff_rate) + "%.")

# The f-string way
print(f"In {year}, {country} had a tariff rate of {tariff_rate}%.")
````

f-strings can also format numbers — decimal places, padding, and more.


````python
tariff_rate = 5.0

print(f"{tariff_rate}")        # whatever Python's default is
print(f"{tariff_rate:.2f}")    # exactly two decimal places
print(f"{tariff_rate:.4f}")    # four decimal places
````

### Lists

A list is an **ordered** collection of items. You write one with square
brackets and commas.


````python
countries = ["Switzerland", "Kenya", "Brazil", "Japan"]
tariff_rates = [1.8, 12.0, 13.5, 4.3]

print(countries)
print(len(countries))     # how many items
print(len(tariff_rates))
````

### Indexing — reaching into a list

You get an item out with square brackets and a **position**. Positions
start at **0**, not 1 — this is confusing at first and then it isn't.


````python
countries = ["Switzerland", "Kenya", "Brazil", "Japan"]

print(countries[0])   # first
print(countries[1])   # second
print(countries[-1])  # last
print(countries[-2])  # second to last
````

### Slicing — reaching for a range

`list[start:stop]` gives you everything from `start` up to *but not
including* `stop`. Leaving either off means "from the beginning" or "to
the end".


````python
countries = ["Switzerland", "Kenya", "Brazil", "Japan", "Argentina"]

print(countries[0:2])   # positions 0 and 1
print(countries[1:4])   # positions 1, 2, 3
print(countries[:3])    # from the start, up to (not including) position 3
print(countries[2:])    # from position 2 to the end
````

The same slicing rules work on strings — a string is a sequence of
characters, after all.


````python
symbol = "WT-DS316-AB-R"

print(symbol[0:2])    # "WT"
print(symbol[3:8])    # "DS316"
print(symbol[-3:])    # "A-R"
````

### Changing a list

Unlike strings, lists **can** be modified in place.


````python
countries = ["Switzerland", "Kenya", "Brazil"]

countries.append("Japan")        # add to the end
print(countries)

countries[0] = "Suisse"          # replace by position
print(countries)

print("Kenya" in countries)      # membership check
print("Chile" in countries)
````

### Dictionaries — labels instead of positions

When you want to look something up by **name** rather than by position,
use a dictionary. It stores **key: value** pairs.


````python
country = {
    "name": "Kenya",
    "population_millions": 55.1,
    "wto_member": True,
    "capital": "Nairobi",
}

print(country["name"])
print(country["capital"])
print(country["population_millions"])
````

Add or update entries by assigning to a key. Ask `in` to check whether
a key exists.


````python
country = {"name": "Kenya", "population_millions": 55.1}

country["region"] = "Africa"                # add a new key
country["population_millions"] = 55.5       # update an existing key

print(country)
print("region" in country)
print("gdp" in country)
````

### Lists of dictionaries — the workhorse shape

Real data almost always looks like this: a list of records, where each
record is a dictionary. This is the shape we'll build the rest of the
course on.


````python
countries = [
    {"name": "Switzerland", "population_millions": 8.8, "avg_tariff": 1.8},
    {"name": "Kenya", "population_millions": 55.1, "avg_tariff": 12.0},
    {"name": "Brazil", "population_millions": 214.3, "avg_tariff": 13.5},
    {"name": "Japan", "population_millions": 125.7, "avg_tariff": 4.3},
]

print(countries[1])                # a whole record
print(countries[1]["name"])        # one field of that record
print(len(countries), "countries loaded")
````

### Tuples and sets, briefly

A **tuple** is a fixed group of values — written with parentheses. Once
made, it can't be changed. It's typically used for something like a
pair of coordinates, or when a function needs to hand you back two
things at once.


````python
bern = (46.94, 7.44)
print(bern[0], bern[1])

# unpacking: two names at once
latitude, longitude = bern
print(latitude, longitude)
````

A **set** holds only unique values. Handy when you want to know
"how many distinct X are there".


````python
tariff_rates = [5.0, 7.5, 5.0, 10.0, 7.5, 5.0]

unique_rates = set(tariff_rates)
print(unique_rates)
print(len(tariff_rates), "values total,", len(unique_rates), "distinct")
````

## Part 3 — Decisions, loops, and functions

*(Live coding — roughly 20 minutes.)*

Everything so far ran once, top to bottom. Now we make code that
decides what to do (`if`), that repeats itself (`for`, `while`), and
that we can call by name (`def`).


### if / elif / else

`if` runs a block only when a condition is `True`. The block is
whatever's **indented** underneath — Python doesn't use braces, it uses
indentation, and it's picky about it. Four spaces is the convention.


````python
tariff_rate = 12.0

if tariff_rate > 10:
    print("above the general threshold")
````

````python
tariff_rate = 5.0

if tariff_rate > 10:
    print("high")
else:
    print("low or moderate")
````

````python
tariff_rate = 7.5

if tariff_rate > 15:
    category = "very high"
elif tariff_rate > 10:
    category = "high"
elif tariff_rate > 5:
    category = "moderate"
else:
    category = "low"

print(category)
````

### for loops — do the same thing to every item

`for` walks through a list (or any other collection) one item at a time.
The variable — `country` in the next cell — takes each value in turn.


````python
countries = ["Switzerland", "Kenya", "Brazil", "Japan"]

for country in countries:
    print("Processing:", country)
````

Anything indented under the `for` runs once per item. Anything not
indented runs only once, after the loop finishes.


````python
countries = ["Switzerland", "Kenya", "Brazil", "Japan"]
count = 0

for country in countries:
    count = count + 1
    print(count, "->", country)

print("Total:", count)
````

### Looping with a counter — `range()` and `enumerate()`

`range(n)` produces the numbers `0, 1, 2, ..., n-1`. Use it when you
need to repeat something *n* times or when you need an index.


````python
for i in range(5):
    print(i, i ** 2)
````

When you want both the index and the item, `enumerate()` gives you
both. This is much cleaner than tracking a counter by hand.


````python
countries = ["Switzerland", "Kenya", "Brazil", "Japan"]

for i, country in enumerate(countries):
    print(f"{i}: {country}")
````

### Looping through a dictionary

Looping over a dictionary gives you its **keys**. `.items()` gives you
key and value together, which is usually what you want.


````python
country = {"name": "Kenya", "population_millions": 55.1, "wto_member": True}

for key, value in country.items():
    print(f"{key} -> {value}")
````

### Combining `if` and `for`

The everyday use of a loop: go through everything, decide something
about each item, keep track of the answer.


````python
countries = [
    {"name": "Switzerland", "avg_tariff": 1.8},
    {"name": "Kenya",       "avg_tariff": 12.0},
    {"name": "Brazil",      "avg_tariff": 13.5},
    {"name": "Japan",       "avg_tariff": 4.3},
]

high_tariff_countries = []

for c in countries:
    if c["avg_tariff"] > 10:
        high_tariff_countries.append(c["name"])

print(high_tariff_countries)
````

### while — repeat until something changes

`while` keeps going as long as its condition is `True`. Handy when you
don't know upfront how many rounds you'll need.


````python
remaining_pages = 137

while remaining_pages > 0:
    print("still", remaining_pages, "pages to read")
    remaining_pages = remaining_pages - 40   # imagine reading 40 pages a day

print("done")
````

### break and continue

- `break` leaves the loop entirely, right now.
- `continue` skips the rest of *this* iteration and goes to the next.


````python
symbols = ["WT-DS316", "notes.txt", "WT-DS135", "draft_v2", "WT-DS58"]

# find the FIRST WT-DS symbol and stop
for s in symbols:
    if s.startswith("WT-DS"):
        print("first hit:", s)
        break
````

````python
symbols = ["WT-DS316", "notes.txt", "WT-DS135", "draft_v2", "WT-DS58"]

# print ALL the WT-DS symbols, skipping everything else
for s in symbols:
    if not s.startswith("WT-DS"):
        continue
    print(s)
````

### Functions — code you can call by name

A function is a piece of code with a name. You **define** it once with
`def`, and then you can **call** it as many times as you want with
different inputs.


````python
def greet(name):
    print(f"Welcome to the course, {name}.")

greet("Ana")
greet("Marco")
greet("Priya")
````

Functions usually **return** a value so the caller can use it, rather
than only printing. `return` hands the value back and exits the
function immediately.


````python
def apply_discount(tariff_rate, discount):
    return tariff_rate * (1 - discount)

new_rate = apply_discount(10.0, 0.20)
print(new_rate)

print(apply_discount(7.5, 0.10))
````

### Default arguments and docstrings

A parameter can have a default value, used when the caller doesn't
supply one. A **docstring**, the string right after `def`, is what
`help()` shows.


````python
def years_since(year_signed, current_year=2026):
    '''Return how many whole years have passed since `year_signed`.'''
    return current_year - year_signed

print(years_since(1994))
print(years_since(1994, current_year=2030))

help(years_since)
````

### Functions can call other functions

This is where the payoff starts — small functions doing one thing each,
composed together.


````python
def classify_tariff(rate):
    if rate > 15:
        return "very high"
    elif rate > 10:
        return "high"
    elif rate > 5:
        return "moderate"
    else:
        return "low"


def describe(country):
    label = classify_tariff(country["avg_tariff"])
    return f"{country['name']}: avg tariff {country['avg_tariff']}% ({label})"


countries = [
    {"name": "Switzerland", "avg_tariff": 1.8},
    {"name": "Kenya",       "avg_tariff": 12.0},
    {"name": "Brazil",      "avg_tariff": 13.5},
]

for c in countries:
    print(describe(c))
````
